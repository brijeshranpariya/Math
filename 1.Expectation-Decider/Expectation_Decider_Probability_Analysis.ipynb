**Objective:** Analyze probability patterns in 200 student records using Python.


## 1. Understanding the Basics

**Probability:** measures how likely an event is to occur.

**Key terms:** Experiment, Outcome, Sample Space, Event, Complement, Intersection (AND), Union (OR), Conditional Probability, Independent Events, Mutually Exclusive Events.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("student_exam_pass_dataset.csv")
df.head()
```

## 2. Empirical and Theoretical Probability

**Empirical:** probability that a randomly selected student passes.

**Theoretical:** assume `Yes` and `No` for group discussion are equally likely.


```python
total = len(df)
pass_count = (df["final_exam_pass"] == "Pass").sum()
empirical_probability = pass_count / total
theoretical_probability = 1 / 2

print("Empirical P(Pass) =", empirical_probability)
print("Theoretical P(Group Discussion = Yes) =", theoretical_probability)
```

## 3. Random Variable & Probability Distribution

Let **X = number of students passing among 3 randomly selected students**.

Possible values: `0, 1, 2, 3`.

```python
from math import comb

p = pass_count / total
q = 1 - p
n = 3
x_values = np.arange(0, 4)
probabilities = [comb(n, x) * p**x * q**(n-x) for x in x_values]

distribution = pd.DataFrame({"X": x_values, "P(X)": probabilities})
distribution
```

```python
mean_x = n * p
variance_x = n * p * q
print("Mean =", mean_x)
print("Variance =", variance_x)
```

## 4. Venn Diagram

A = `study_hours > 10`

B = `attendance > 80`

```python
A = df["study_hours"] > 10
B = df["attendance"] > 80

a_only = (A & ~B).sum()
intersection = (A & B).sum()
b_only = (~A & B).sum()
neither = (~A & ~B).sum()

print("A only:", a_only)
print("A AND B:", intersection)
print("B only:", b_only)
print("Neither:", neither)
```

## 5. Contingency Table & Probability Calculations

```python
table = pd.crosstab(df["group_discussion"], df["final_exam_pass"], margins=True)
table
```

```python
yes_pass = ((df["group_discussion"] == "Yes") & (df["final_exam_pass"] == "Pass")).sum()
yes_count = (df["group_discussion"] == "Yes").sum()

joint_probability = yes_pass / total
marginal_probability = pass_count / total
conditional_probability = yes_pass / yes_count

print("Joint probability =", joint_probability)
print("Marginal probability =", marginal_probability)
print("Conditional probability =", conditional_probability)
```

## 6. Understanding Relationships

Conditional probability: among students who participated in group discussion, what proportion passed?

For independence: `P(A ∩ B) = P(A)P(B)`.

```python
p_discussion = yes_count / total
p_pass = pass_count / total
expected_joint = p_discussion * p_pass

print("P(Discussion = Yes) =", p_discussion)
print("P(Pass) =", p_pass)
print("P(Discussion = Yes) × P(Pass) =", expected_joint)
print("Actual joint probability =", joint_probability)
print("Conclusion:", "Independent" if np.isclose(joint_probability, expected_joint) else "Dependent")
print("Mutually exclusive:", "Yes" if yes_pass == 0 else "No")
```

## 7. Bayes Theorem

Given:
- `P(High Attendance | Pass) = 0.70`
- `P(High Attendance | Fail) = 0.40`
- `P(High Attendance) = 0.60`

Find `P(Pass | High Attendance)`.

```python
P_H_given_P = 0.70
P_H_given_F = 0.40
P_H = 0.60

P_Pass = (P_H - P_H_given_F) / (P_H_given_P - P_H_given_F)
P_Pass_given_H = (P_H_given_P * P_Pass) / P_H

print("P(Pass) =", P_Pass)
print("P(Pass | High Attendance) =", P_Pass_given_H)
print("Percentage =", P_Pass_given_H * 100, "%")
```
